```sql
describe ecomotive.gold.fact_lecturas_iot
```

```sql


MERGE INTO ecomotive.gold.fact_lecturas_iot as target
USING (

    select 
            s.row_hash,
            ds.sk_sensor,
            dv.sk_vehiculo,
            df.fecha_key,
            s.momento_de_lectura,
            s.temperatura_c,
            s.latitud,
            s.longitud,
            s.porcentaje_bateria_sensor,
            s.signal_strength_dbm,
            s.estado,
            s._record_added_at AS _added_at
            
    from ecomotive.silver.validos_iot_sensors s
    left join ecomotive.gold.dim_sensores ds on s.sensor_id = ds.sensor_id
    left join ecomotive.gold.dim_vehiculos dv on s.vehiculo_id = dv.vehiculo_id
    left join ecomotive.gold.dim_fecha df on TO_DATE(s.momento_de_lectura) =  df.fecha ) as source

ON target.row_hash = source.row_hash

WHEN  NOT MATCHED THEN
 INSERT (
        row_hash,
        sk_sensor,
        sk_vehiculo,
        fecha_key,
        momento_de_lectura,
        temperatura_c,
        latitud,
        longitud,
        porcentaje_bateria_sensor,
        signal_strength_dbm,
        estado,
        _added_at
    )
    VALUES (
        source.row_hash,
        source.sk_sensor,
        source.sk_vehiculo,
        source.fecha_key,
        source.momento_de_lectura,
        source.temperatura_c,
        source.latitud,
        source.longitud,
        source.porcentaje_bateria_sensor,
        source.signal_strength_dbm,
        source.estado,
        source._added_at
    );
```
