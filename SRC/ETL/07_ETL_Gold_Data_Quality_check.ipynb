### 1. Integridad de las dimensiones

```sql

SELECT sensor_id, COUNT(*) AS cantidad
FROM ecomotive.gold.dim_sensores
GROUP BY sensor_id
HAVING COUNT(*) > 1;
```

```sql
SELECT
Count(*) as cantidad,
vehiculo_id
FROM ecomotive.gold.dim_vehiculos
GROUP BY vehiculo_id
HAVING COUNT(*) > 1;
```

### 2. Integridad referencial de FACT

```sql
SELECT COUNT(*) AS registros_sin_sensor
FROM ecomotive.gold.fact_lecturas_iot f
LEFT JOIN ecomotive.gold.dim_sensores d
    ON f.sk_sensor = d.sk_sensor
WHERE d.sk_sensor IS NULL;
```

### 3. Unicidad de eventos

```sql

SELECT row_hash, COUNT(*) AS cantidad
FROM ecomotive.gold.fact_lecturas_iot
GROUP BY row_hash
HAVING COUNT(*) > 1;
```

### 4 Consistencia entre Silver y Gold

```sql
SELECT
    (SELECT COUNT(*)
     FROM ecomotive.silver.validos_iot_sensors) AS total_silver,

    (SELECT COUNT(*)
     FROM ecomotive.gold.fact_lecturas_iot) AS total_gold;
```
