```sql
CREATE OR REPLACE VIEW ecomotive.views.vw_lecturas_iot AS

SELECT
    -- Lectura
    f.momento_de_lectura,
    f.temperatura_c,
    f.porcentaje_bateria_sensor,
    f.signal_strength_dbm,
    f.estado,

    -- Sensor
    s.sensor_id,
    s.tipo_sensor,
    s.fabricante,
    s.fecha_instalacion,
    s.firmware_version,

    -- Vehículo
    v.vehiculo_id,
    v.placa,
    v.marca,
    v.modelo,
    v.tipo_vehiculo

FROM ecomotive.gold.fact_lecturas_iot f

LEFT JOIN ecomotive.gold.dim_sensores s
    ON f.sk_sensor = s.sk_sensor

LEFT JOIN ecomotive.gold.dim_vehiculos v
    ON f.sk_vehiculo = v.sk_vehiculo;
```
